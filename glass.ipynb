```python
options = Options()
options.headless = True

# Initialize webdriver
driver = webdriver.Chrome(options=options)

# List to store extracted data

product_badge = []
product_colors = []
product_code = []
product_brand = []
product_price = []


# Assuming knowledge of total pages
total_pages = 10

for page_number in range(1, total_pages + 1):
    url = f"https://www.glasses.com/gl-us/eyeglasses?pageNumber={page_number}"
    driver.get(url)
    time.sleep(10)

# After stabilizing the page, we now use Beautifulsoup to parse the loaded page_source

    soup = BeautifulSoup(driver.page_source, 'html.parser')

# Now use the class selector to retrieve the glass card
    glasses = soup.find_all("a", class_="product-tile")

    print(f"Page {page_number}: {len(glasses)} product cards")


    for glass in glasses:
        badges = glass.find_all("div", class_="product-badge")
        colors = glass.find("div", class_="product-colors")
        code = glass.find("div", class_="product-code")
        brand = glass.find("div", class_="product-brand")
        price = glass.find("div", class_="product-list-price")
        
        
        product_badge.append(",".join(badge.get_text(strip=True) for badge in badges) or None)
        
        product_colors.append(colors.get_text(strip=True) if colors else None)
        
        product_code.append(code.get_text(strip=True) if code else None)
        
        product_brand.append(brand.get_text(strip=True) if brand else None)
        
        product_price.append(price.get_text(strip=True) if price else None)
        

# Get the page source and close the browser
# page_source = driver.page_source
driver.quit()
```

```python
data = {
"product_badge" : product_badge,
"product_colors" : product_colors,
"product_code" : product_code,
"product_brand": product_brand,
"product_price": product_price
}


glass_df = pd.DataFrame(data)
```

```python

# Convert price to float
glass_df['product_price'] = glass_df['product_price'].str.replace('$', '', regex=False).str.strip().astype('float')
```
